```python
from pathlib import Path
import pandas as pd

station_file = Path("data/raw/stations/hly3723_subset.csv")

# The first 41 lines describe the station; line 42 contains column names.
station = pd.read_csv(station_file, skiprows=41, nrows=5)

print("Columns:", station.columns.tolist())
display(station)
```

```python
# Read all hourly records for this station
station = pd.read_csv(station_file, skiprows=41)

print("Number of rows:", len(station))
print("Column names:", station.columns.tolist())

# Look at the date, rainfall, and its indicator
display(station[["date", "ind", "rain"]].head(10))
print("Rainfall indicator values:")
print(station["ind"].value_counts(dropna=False))
```

```python
print("Rainfall indicator counts:")
print(station["ind"].value_counts(dropna=False))

print("\nRainfall values:")
print(station["rain"].describe())

print("\nHours with rainfall above zero:")
print((station["rain"] > 0).sum())
```

```python
station_files = [
    "hly3904_subset.csv",
    "hly518_subset.csv",
    "hly4935_subset.csv",
    "hly532.csv",
]

for filename in station_files:
    path = Path("data/raw/stations") / filename
    other_station = pd.read_csv(path, skiprows=41)

    # Dublin's file also contains years outside 2024
    rows_2024 = other_station[
        other_station["date"].str.contains("-2024 ", na=False)
    ]

    print(f"\n{filename}: {len(rows_2024)} rows in 2024")
    print(rows_2024["ind"].value_counts(dropna=False))
```

```python
dublin_file = Path("data/raw/stations/hly532.csv")

# Read only the columns we need; this also avoids the mixed-type warning
dublin = pd.read_csv(
    dublin_file,
    skiprows=41,
    usecols=["date", "ind", "rain"]
)

for year in [2023, 2024]:
    rows = dublin[dublin["date"].str.contains(f"-{year} ", na=False)]
    print(f"\nDublin {year}: {len(rows)} rows")
    print(rows["ind"].value_counts(dropna=False).head())
```

```python
station["date"] = pd.to_datetime(
    station["date"],
    format="%d-%b-%Y %H:%M",
    utc=True
)

print("First timestamp:", station["date"].min())
print("Last timestamp:", station["date"].max())
print("Number of timestamps:", station["date"].nunique())

display(station[["date", "rain", "ind"]].head())
```

```python
weather_columns = ["temp", "rhum", "msl", "clamt"]

display(station[["date"] + weather_columns].head(10))

print("Missing values:")
print(station[weather_columns].isna().sum())

print("\nTemperature indicator:")
print(station["ind.1"].value_counts(dropna=False))
```

```python
import numpy as np

# Find the cloud file inside the data folder
cloud_file = next(Path("data").rglob("ireland_clouds_*.npz"))

# Load only its timestamps
with np.load(cloud_file) as cloud_data:
    cloud_times = pd.to_datetime(cloud_data["times"], utc=True)

print("Cloud hours:", len(cloud_times))
print("Station hours:", len(station))
print("Cloud period:", cloud_times.min(), "to", cloud_times.max())
print("Station period:", station["date"].min(), "to", station["date"].max())

matching_hours = station["date"].isin(cloud_times).sum()
print("Station hours matching cloud hours:", matching_hours)
```

```python
# One row for each cloud image time
cloud_hours = pd.DataFrame({"date": cloud_times})

# Add Casement measurements for the same hour
casement_aligned = cloud_hours.merge(
    station[["date", "rain", "ind", "temp", "rhum", "msl", "clamt"]],
    on="date",
    how="left",
    validate="one_to_one"
)

print("Aligned rows:", len(casement_aligned))
display(casement_aligned.head())
```

```python
print("Aligned rows:", len(casement_aligned))
print("\nMissing values after alignment:")
print(casement_aligned.isna().sum())
```

```python
station_names = {
    "Casement": "hly3723_subset.csv",
    "Cork Airport": "hly3904_subset.csv",
    "Shannon Airport": "hly518_subset.csv",
    "Knock Airport": "hly4935_subset.csv",
    "Dublin Airport": "hly532.csv",
}

columns_to_check = ["date", "rain", "temp", "rhum", "msl", "clamt"]

for name, filename in station_names.items():
    path = Path("data/raw/stations") / filename

    data = pd.read_csv(path, skiprows=41, usecols=columns_to_check)
    data = data[data["date"].str.contains("-2024 ", na=False)]

    print(f"\n{name}: {len(data)} hours")
    print(data.isna().sum())
```

```python
measurements = ["rain", "temp", "rhum", "msl", "clamt"]

for name, filename in station_names.items():
    path = Path("data/raw/stations") / filename

    data = pd.read_csv(
        path,
        skiprows=41,
        usecols=["date"] + measurements,
        low_memory=False
    )
    data = data[data["date"].str.contains("-2024 ", na=False)]

    # Unreadable numbers become NaN, so we can count them
    numeric_data = data[measurements].apply(pd.to_numeric, errors="coerce")

    print(f"\n{name} — missing or non-numeric values:")
    print(numeric_data.isna().sum())
```

```python
for name, filename in station_names.items():
    path = Path("data/raw/stations") / filename
    dates = pd.read_csv(path, skiprows=41, usecols=["date"])["date"]
    dates = dates[dates.str.contains("-2024 ", na=False)]
    dates = pd.to_datetime(dates, format="%d-%b-%Y %H:%M", utc=True)

    missing_hours = (~cloud_times.isin(dates)).sum()
    repeated_hours = dates.duplicated().sum()

    print(f"{name}: {missing_hours} missing hours, {repeated_hours} repeated hours")
```
